### Bucle `while`
* Repetición de un bloque de código hasta que se deje de cumplir una expresión (es decir, hasta que una condición evalúe a `False`).
* Si la condición evalúa a `False` desde el principio, el bloque de código nunca se ejecuta.
* Cuidado con los bucles infinitos.

Formato general:
```text
while test:         # Mientras se cumple la condición 
    statements      # instrucciones a ejecutar

````python
# # Ejemplo: Mostrar los primeros 3 objetos de una lista

indice = 0
numeros = [9, 4, 7, 1, 2]

while indice < 3:
    print(numeros[indice])
    indice =indice+1
````

````python
# Ejemplo: contar y mostrar los números inferiores a 10.
numeros = [33, 3, 9, 21, 1, 7, 12, 10, 8]
contador = 0
indice = 0
while indice < len(numeros):
    if numeros[indice] < 10:
        print(numeros[indice])
        contador += 1
    indice += 1
    
print('Contador:', contador)
print('Indice:', indice)
````

````python
nombre = 'Pablo'
while nombre: # Mientras 'nombre' no sea vacío
    print(nombre)
    nombre = nombre[1:]
````

````python
# #bucle infinito
i = 0
while i < 10:
    print(i)
    i += 1
````

### 3.1 Las sentencias `break`, `continue` y `else`

#### Break y Continue
* Solo tienen sentido dentro de bucles.
* `break` permite terminar el bucle por completo.
* `continue` permite saltar a la siguiente iteración, continuando con el bucle.
* Pueden aparecer en cualquier parte de un bucle, pero normalmente aparecen dentro de sentencias condicionales (`if`).

````python
# # Ejemplo: break
rating_to_find = 4.2
movie_ratings = [4.3, 2.5, 1.7, 4.2, 3.8, 3.3, 4.5]
for rating in movie_ratings:
    print(rating)
    if rating == rating_to_find:
        print("Found")
        break
        
print(rating)
````

````python
for i in range(10):
    if i % 2 == 0:
        continue 
# Si el número es par, salta a la siguiente iteración
print(f'Odd number {i}')
````

* Observa como la sentencia continue te puede ayudar a reducir el número de niveles de
anidamiento.
* Sin continue el anterior ejemplo sería:


````python
for i in range(10):
    if i % 2 != 0:
        print('Numero impar:', end=' ')
        print(i)
````

````python
for i in range(5):
    for a in range(2):
        print(f"{i} - {a}")
        if i == 3 and a == 0:
            break
````

#### Else 
* Los bucles pueden tener una sensencia else.
* Resulta poco intuitiva para muchos programadores porque esta sintaxis no existe en otros
lenguajes.
* Se ejecuta cuando el bucle termina con normalidad; es decir, cuando no termina a causa de
un *break*.

````python
lista = [60,60,30,60]
element_to_find = 60

for element in lista:
    if element == element_to_find:
        print("found")
        break
else:
    print("not found")
````

````python
# with flag
lista = [10,30,50,40]
element_to_find = 30
found = False

for element in lista:
    if element == element_to_find:
        found = True
        break
if found:
    print("dato encontrado")
else:
    print("not found")
````

## 4 List Comprehensions
* Permiten construir listas a través de la ejecución repetida (sentencia `for`) de una expresión para cada ítem de un objeto iterable.
* Van entre `[` y `]`. Esto es indicativo de que estamos construyendo una lista.

Sintaxis básica:
```text
[<expression> for <item> in <iterable>]

Ejemplo: repetir los caracteres de un string

````python
[char*3 for char in "Enrique"]
````

````python
lista = []
for char in 'Enrique':
    lista.append(char)
print(lista)
````

Comúnmente, el item de la sentencia *for* aparecerá en la expresión principal, pero eso no es obli￾gatorio.


````python
[2 for _ in 'Enrique']
````

### 4.0.1 Versión extendida 
Se puede especificar un filtro (sentencia `if`) para obtener únicamente los elementos que cumplan cierta condición.

Sintaxis:
```text
[<expression> for <item> in <iterable> if <condition>]
Ejemplo: obtener números pares:


````python
[y for y in range(9) if y % 2 == 0]
````

* Las list comprehension no son realmente requeridas, ya que siempre podemos escribir un bucle
equivalente.


````python
pares = []
for x in range(9):
    if x % 2 == 0:
        pares.append(x)
print(pares)
````

### 4.0.2 Versión completa (`if-else` interno)
La sentencia `if` de una list comprehension también puede contener una expresión alternativa.

Sintaxis:
```text
[<expression_1> if <condition> else <expression_2> for <item> in <iterable>]

Ejemplo: poner a cero los números pares.


````python
[0 if x % 2 == 0 else x for x in range(9)]
````

````python
[x**2 if x > 2 else x for x in range(-4,5) if x > 0]
````

````python
lista = []
for x in range(-4,5):
    if x > 0:
        if x > 2:
            lista.append(x**2)
        else:
            lista.append(x)
print(lista)
````

````python
[x**2 if x > 0 else 100 if x==0 else x for x in range(-4,5)]
````

````python
lista = []
for x in range(-4,5):
    if x > 0:
        lista.append(x**2)
    elif x == 0:
        lista.append(100)
    else:
        lista.append(x)
print(lista)
````

### 4.0.3 Anidamiento
Las list comprehensions soportan anidamiento en sus expresiones.

**Ejemplo:** bucle anidado para obtener una lista de tuplas que combinan los elementos de dos listas dadas.

````python
lista_1 = [1, 2, 3]
lista_2 = [4, 5]
[(x, y) for x in lista_1 for y in lista_2]
````

### 4.0.4 Pros y contras
* Principales ventajas de las comprehensions en comparación a un bucle convencional:
    * Expresión compacta y legible, si estás familiarizado con la sintaxis.
    * Mejor rendimiento.
* Desventaja: no escalan bien. Una list comprehension se puede convertir rápidamente en una
expresión difícil de entender.


### 4.0.5 Otros tipos de comprehensions

Usando `{` y `}` como delimitadores y una expresión `clave: valor`, se obtiene un diccionario en

````python
d = {x : x*x for x in range(10)}
print(d)
print(type(d))
````

````python
items = ['Banana', 'Pear', 'Olives']
price = [1.1, 1.4, 2.4]
shopping = {k:v for k,v in zip(items,price)}
print(shopping)
````

````python
# Ejercicio 1
# Definimos la lista inicial con elementos puramente numéricos [cite: 583, 616]
numeros = [1, 5, 9, 2, 3]

# Inicializamos una variable acumuladora en 0 para almacenar el resultado [cite: 617]
suma = 0

# Iteramos de forma secuencial sobre cada elemento de la lista 'numeros' [cite: 618]
for numero in numeros:
    # Sumamos el valor del elemento actual a nuestra variable acumuladora [cite: 619]
    suma += numero  # Esto equivale a escribir: suma = suma + numero

# Imprimimos en la terminal el valor total acumulado tras finalizar el ciclo [cite: 620]
print(suma)
````

````python
# Ejercicio 2
# Definimos la lista inicial que contiene múltiples valores repetidos [cite: 585, 623]
numeros = [1, 2, 2, 3, 4, 5, 4, 5, 3, 3, 1]

# Creamos una lista vacía donde guardaremos los elementos únicos de forma ordenada [cite: 624]
resultado = []

# Recorremos uno a uno todos los números presentes en la lista original [cite: 626]
for numero in numeros:
    # Evaluamos si el número actual NO ha sido añadido aún a la lista de resultados [cite: 627]
    if numero not in resultado:
        # Si no está presente, lo añadimos al final de la lista 'resultado' [cite: 628]
        resultado.append(numero)

# Mostramos la nueva lista final que ya no contiene ningún dato duplicado [cite: 629]
print(resultado)
````

````python
# Ejercicio 3
# Definimos el límite superior 'n' para la generación de elementos [cite: 587, 632]
n = 5

# Inicializamos un diccionario vacío donde guardaremos las parejas clave-valor [cite: 633]
diccionario = {}

# Iniciamos un ciclo que va desde 1 hasta 'n' inclusive (por eso se usa n+1) [cite: 634]
for i in range(1, n + 1):
    # Asignamos el cuadrado del número actual 'i' a la clave 'i' en el diccionario [cite: 635]
    diccionario[i] = i * i

# Imprimimos la estructura del diccionario completo obtenido en pantalla [cite: 636]
print(diccionario)
````

````python
# Ejercicio 4
# Lista de strings con diferentes longitudes y caracteres iniciales [cite: 638]
palabras = ["", "Valencia", "python", "asignaturas", "programación", "java", "estudiantes", "académicos"]

# Recorremos secuencialmente cada string dentro de la lista de palabras [cite: 639]
for palabra in palabras:
    # Validamos dos condiciones: que mida 9 o más caracteres Y que inicie con la letra 'a' [cite: 640]
    if len(palabra) >= 9 and palabra.startswith('a'):
        # Si cumple ambas condiciones, imprimimos un mensaje mostrando la palabra [cite: 640]
        print(f"Palabra encontrada: {palabra}")
        # Rompemos el ciclo inmediatamente para finalizar tras el primer hallazgo [cite: 590, 640]
        break
# El bloque 'else' asociado al 'for' se ejecuta solo si el bucle terminó sin activar ningún 'break' [cite: 371, 641]
else:
    # Informamos al usuario que ninguna palabra de la lista cumplió los requisitos [cite: 642]
    print("Palabra no encontrada")
````

````python
#Ejercicio 5
# Lista de números enteros positivos de entrada [cite: 593, 644]
L = [3, 5, 13, 12, 1, 9]

# Creamos una lista vacía para ir almacenando las posiciones válidas encontradas [cite: 645]
resultado = []

# Iteramos numéricamente a través de un rango basado en el tamaño total de la lista 'L' [cite: 257, 646]
for i in range(len(L)):
    # Evaluamos si el elemento situado en el índice 'i' tiene residuo cero al dividirse por 3 [cite: 647]
    if L[i] % 3 == 0:
        # Si es múltiplo de 3, agregamos el ÍNDICE (la posición 'i') a la lista [cite: 648]
        resultado.append(i)

# Mostramos en consola la lista resultante con los índices recolectados [cite: 649]
print(resultado)
````

````python
# Ejercicio 6
# Definimos el diccionario con claves string y valores numéricos de prueba [cite: 594, 651]
d = {'a': 4.3, 'b': 1, 'c': 7.8, 'd': -5}

# Inicializamos el valor máximo en infinito negativo para asegurar que cualquier número real sea mayor [cite: 653]
maximo = float("-inf")

# Inicializamos la variable que guardará la clave ganadora en un estado vacío [cite: 660]
resultado = None

# Verificamos de forma preventiva si el diccionario no tiene elementos [cite: 655]
if len(d) == 0:
    print("El diccionario está vacío") [cite: 656]
else:
    # Recorremos el diccionario obteniendo simultáneamente cada clave y su valor numérico [cite: 293, 657]
    for clave, valor in d.items():
        # Comparamos si el valor del elemento actual supera al máximo registrado previamente [cite: 658]
        if valor > maximo:
            # Actualizamos nuestro registro del valor más alto encontrado hasta ahora [cite: 659]
            maximo = valor
            # Guardamos el nombre de la clave asociada a dicho valor numérico máximo [cite: 660]
            resultado = clave
            
    # Imprimimos la clave correspondiente al número más grande del diccionario [cite: 661]
    print(resultado)
````

````python
# Ejercicio 7
# Definimos las dos listas numéricas independientes de entrada [cite: 596, 663, 664]
a = [2, 4, 6, 8]
b = [7, 11, 15, 22]

# Iniciamos el bucle externo para recorrer individualmente los elementos de la lista 'a' [cite: 665]
for i in a:
    # Filtramos para que únicamente continúen los números mayores estrictos que 5 [cite: 596, 666]
    if i > 5:
        # Iniciamos el bucle interno (anidado) para recorrer los elementos de la lista 'b' [cite: 667]
        for j in b:
            # Filtramos para procesar solo aquellos números que sean menores que 14 [cite: 596, 668]
            if j < 14:
                # Imprimimos detalladamente la operación matemática y su resultado en pantalla [cite: 673]
                print(f"{i} x {j} = {i * j}")
````

````python
# Ejercicio 8
# Iniciamos un bloque de pruebas para capturar posibles errores en tiempo de ejecución [cite: 518, 670]
try:
    # Solicitamos la entrada por consola del usuario (retorna un tipo string) [cite: 600, 671]
    x = input("Introduce un número para dividir 10 por él: ")
    
    # Intentamos convertir la entrada string a un tipo flotante procesable matemáticamente [cite: 674]
    numero = float(x)
    
    # Realizamos la operación matemática e imprimimos el resultado directamente [cite: 675]
    print(f"El resultado de 10/{x} es: {10 / numero}")

# Capturamos el error en caso de que el usuario haya ingresado texto no numérico [cite: 676]
except ValueError:
    # Mostramos un mensaje advirtiendo que el formato ingresado es incorrecto [cite: 677]
    print(f"El valor '{x}' no es numérico")

# Capturamos el error específico que ocurre si el usuario ingresa el número cero [cite: 678]
except ZeroDivisionError:
    # Mostramos un mensaje indicando que la división por cero no está definida [cite: 679]
    print("Error: división por cero")
````

````python
# Ejercicio 9
# Definimos un diccionario base asignando notas académicas a asignaturas [cite: 681]
notas = {'Programación': 9.5, "Inteligencia artificial": 7.6, "Redes": 5.2, "Sistemas operativos": 6.4}

# Estructuramos un bloque try para mitigar búsquedas de elementos inexistentes [cite: 518, 682]
try:
    # Solicitamos al usuario que escriba el nombre exacto de la materia a consultar [cite: 602, 683]
    asignatura = input("Introduce una asignatura: ")
    
    # Intentamos acceder al valor de la clave provista y mostrar su respectiva nota [cite: 603, 684]
    print(f"La nota de la asignatura '{asignatura}' es: {notas[asignatura]}")

# Capturamos el error específico de Python cuando no halla la clave dentro del diccionario [cite: 685]
except KeyError:
    # Informamos formalmente al usuario que la materia consultada no está registrada [cite: 604, 686]
    print(f"Asignatura incorrecta: {asignatura}")
````

````python
# Ejercicio 10
# Lista original que combina enteros y flotantes, tanto positivos como negativos [cite: 605, 606, 688]
lista = [1, 4, -3, -1.5, 6.5, 2, 8, 2.1]

# Expresión compacta que filtra la lista:
# 1. 'for numero in lista' recorre la colección[cite: 400].
# 2. 'type(numero) == int' valida que no sea float[cite: 606, 690].
# 3. 'numero > 0' garantiza que sea positivo[cite: 605, 690].
resultado = [numero for numero in lista if type(numero) == int and numero > 0]

# Mostramos el resultado de la comprensión (generará: [1, 4, 2, 8])
print(resultado)
````

````python
# Ejercicio 11
# Definimos el string de evaluación [cite: 607, 692]
palabra = "murcielago"

# Expresión compacta usando llaves {} para obtener un conjunto (evita duplicados de forma nativa):
# 1. Recorremos letra por letra la palabra[cite: 266, 693].
# 2. 'if letra in (...)' filtra dejando pasar solo las vocales declaradas[cite: 693].
vocales = {letra for letra in palabra if letra in ('a', 'e', 'i', 'o', 'u')}

# Mostramos el conjunto de vocales resultantes
print(vocales)
````

````python
# Ejercicio 12
# Expresión compacta que evalúa el rango numérico de 0 a 50 (range(51)):
# 1. 'for numero in range(51)' genera los números de manera secuencial[cite: 245, 695].
# 2. 'str(numero)' convierte temporalmente el número a texto para poder buscar caracteres individuales[cite: 695].
# 3. "'3' in str(numero)" comprueba si el carácter del dígito 3 forma parte del texto[cite: 695].
numeros_con_tres = [numero for numero in range(51) if '3' in str(numero)]

# Mostramos la lista con los números que cumplen la condición [cite: 608]
print(numeros_con_tres)
````

````python
# Ejercicio 13 
# Definimos la frase de entrada que vamos a procesar [cite: 610, 697]
frase = "Soy un ser humano"

# Expresión compacta para generar un diccionario mapeado:
# 1. 'frase.split()' fragmenta la oración convirtiéndola en una lista de palabras individuales usando los espacios[cite: 698].
# 2. 'for palabra in ...' pasa por cada una de estas palabras obtenidas[cite: 698].
# 3. 'palabra: len(palabra)' define la palabra actual como clave y su longitud entera como valor[cite: 609, 698].
diccionario_longitudes = {palabra: len(palabra) for palabra in frase.split()}

# Imprimimos el diccionario resultante con los nombres de las palabras y sus tamaños [cite: 699]
print(diccionario_longitudes)
````

````python
# Ejercicio 14
# Definimos un diccionario de traducción para la conversión de la segunda mitad del rango [cite: 701, 703]
numero_a_palabra = {6: "seis", 7: "siete", 8: "ocho", 9: "nueve", 10: "diez"}

# Expresión extendida completa:
# 1. 'for numero in range(1, 11)' genera la sucesión del 1 al 10[cite: 704].
# 2. Evaluamos la condición ternaria interna para cada número generado[cite: 434]:
#    - Si el número es menor o igual a 5, preservamos el número entero intacto[cite: 704].
#    - De lo contrario (del 6 al 10), extraemos su traducción textual desde el diccionario[cite: 704].
resultado_mixto = [numero if numero <= 5 else numero_a_palabra[numero] for numero in range(1, 11)]

# Mostramos la lista combinada final generada con éxito [cite: 706]
print(resultado_mixto)
````
