## 1 Funciones

### 1.1 ¿Qué son?
* Son un mecanismo que te ofrece el lenguaje para agrupar conjuntos de instrucciones de manera que se puedan ejecutar cuando el programador considere oportuno.
* Las funciones van identificadas por un nombre.
* Se debe usar este nombre para invocar (ejecutar) el código de la función.
* Las funciones calculan un resultado (output) en base a unos parámetros de entrada (input).
* En cada invocación de la función, los parámetros de entrada pueden variar.

## 1.2 Ejemplo

## 1.4 Programación de funciones en Python


## 1.5 Paso de parámetros
* Tipos simples (inmutables) por valor: int, float, string.
    * El efecto es como si se creara una copia dentro de la función, aunque realmente no es
esto lo que ocurre.
    * Los cambios dentro de la función no afectan fuera.

````python
def cambiar_valor(x):
    x += 3
    print('Dentro de la función:', x) # x ahora referencia a un nuevo objeto␣(el 3), pero esto no afecta a la variable de fuera de la función, que sigue␣apuntando al 2.
    return
a = 2
cambiar_valor(a)
print('Fuera de la función:', a)
````

* Tipos complejos (mutables) por **referencia**: list, set, dictionary.
    * Dentro de la función se maneja el mismo objeto que se ha pasado desde fuera.
    * Los cambios dentro de la función sí afectan fuera

````python
def anyadir_2_a(x):
    x.append(4)
lista = [0, 1,3]
anyadir_2_a(lista)
print(lista)
````

````python
# with tuples
def anyadir_2_a(x):
    print(id(x))
    return
lista = (0, 1)
print(id(lista))
anyadir_2_a(lista)
print(lista)
````

## 1.9 Scopes

* La localización de una asignación a una variable en el código determina desde donde puedes
acceder a esa variable.
* Por ejemplo, una variable asignada dentro de una función sólo es visible dentro de esa función.
* El alcance de la visibilidad de una variable determina su scope.
* El término scope hace referencia a un espacio de nombres (namespace). Por ejemplo, una
función establece su propio namespace.

````python
# X = 10
# def func():
# X = 20 # Local a la función. Es una variable diferente, no visible␣fuera de 'func'
# def func_int():
# X = 30
# print(X)
# func_int()
# func()
````

## Regla LEGB 

Dada una función *F*, tenemos los siguientes scopes:

* **L**ocal: variables locales a F.
* **E**nclosing: variables localizadas en funciones que contienen a (o por encima de) F.
* **G**lobal: variables definidas en el módulo (fichero) que no están contenidas en ninguna función.
    Este scope no abarca más de un fichero.
* **B**uilt-ins: proporcionadas por el lenguaje.
  
La resolución de nombres ocurre de abajo a arriba.

## 1.11 Resolucion de Ejercicios



````python
#Ejercicio 1
# Definición de la función para calcular los cuadrados de una lista
def calcular_cuadrados(lista_numeros):
    # Creamos una lista vacía para almacenar los resultados
    resultados = []
    # Iteramos a través de cada número en la lista de entrada
    for numero in lista_numeros:
        # Elevamos el número actual al cuadrado y lo añadimos a la lista
        resultados.append(numero ** 2)
    # Retornamos la lista final con los resultados
    return resultados

# --- PRUEBA DEL EJERCICIO 1 ---
lista_prueba = [1, 5, 9, 2, 3]
print("Lista original:", lista_prueba)
print("Resultado cuadrados:", calcular_cuadrados(lista_prueba))
````

````python
# Ejercicio 2
# Definición de la función con argumentos indeterminados (*args)
def sumar_numeros_indeterminados(*numeros):
    # Inicializamos el acumulador de la suma en 0
    suma_total = 0
    # Iteramos sobre la tupla de argumentos empaquetados
    for numero in numeros:
        # Sumamos el número actual al total
        suma_total += numero
    # Retornamos el resultado de la sumatoria
    return suma_total

# --- PRUEBA DEL EJERCICIO 2 ---
# Pasamos los números directamente como parámetros independientes (sin listas)
resultado_suma = sumar_numeros_indeterminados(10, 20, 30, 5)
print("Resultado de la suma indeterminada:", resultado_suma)
````

````python
# Ejercicio 3
# Definición de la función convencional para invertir una cadena
def invertir_string(cadena):
    # Retornamos el string invertido utilizando la técnica de slicing [::-1]
    return cadena[::-1]

# --- PRUEBA DEL EJERCICIO 3 ---
texto = "hola gente"
print(f"Texto original: '{texto}'")
print(f"Texto invertido: '{invertir_string(texto)}'")
````

````python
# Ejercicio 4
# Definición de la función lambda anónima para invertir una cadena
invertir_string_lambda = lambda cadena: cadena[::-1]

# --- PRUEBA DEL EJERCICIO 4 ---
texto_lambda = "hola Inge"
print(f"Texto original (Lambda): '{texto_lambda}'")
print(f"Texto invertido (Lambda): '{invertir_string_lambda(texto_lambda)}'")
````

````python
# Ejercicio 5
# Definición de la función para comprobar límites de un rango
def comprobar_en_rango(numero, limite_inferior, limite_superior):
    # Evaluamos mediante una comparación encadenada si se encuentra dentro de los límites
    if limite_inferior <= numero <= limite_superior:
        # Retornamos True si el número está dentro del rango especificado
        return True
    else:
        # Retornamos False si el número está fuera del rango
        return False

# --- PRUEBA DEL EJERCICIO 5 ---
print("¿El 5 está entre 1 y 10?:", comprobar_en_rango(5, 1, 10))   
print("¿El 12 está entre 1 y 10?:", comprobar_en_rango(12, 1, 10)) 
````

````python
# Ejercicio 6
# Definición de la función para obtener los primeros 5 múltiplos
def obtener_cinco_multiplos(numero):
    # Validamos si el parámetro recibido es menor o igual a cero (incorrecto)
    if numero <= 0:
        # Mostramos el mensaje de error solicitado por pantalla
        print("Error: El número ingresado debe ser un entero positivo mayor a cero.")
        # Retornamos una lista vacía inmediatamente
        return []
    
    # Si el parámetro es correcto, inicializamos la lista de múltiplos
    multiplos = []
    # Usamos un bucle for de 1 a 5 para calcular los múltiplos secuenciales
    for i in range(1, 6):
        # Multiplicamos el número base por el iterador actual y lo añadimos a la lista
        multiplos.append(numero * i)
    # Retornamos la lista final de múltiplos calculados
    return multiplos

# --- PRUEBA DEL EJERCICIO 6 ---
print("Múltiplos de 3:", obtener_cinco_multiplos(3))   
print("Múltiplos de -5:", obtener_cinco_multiplos(-5))
````

````python
# Ejercicio 7
# Definición de la función para verificar duplicados usando sets
def verificar_duplicados(lista):
    # Si la longitud original es diferente a la del set, significa que había duplicados
    if len(lista) != len(set(lista)):
        # Retornamos True si contiene duplicados
        return True
    else:
        # Retornamos False si todos los elementos son únicos
        return False

# --- PRUEBA DEL EJERCICIO 7 ---
lista_con_duplicados = [1, 2, 3, 2, 4]
lista_sin_duplicados = [1, 2, 3, 4, 5]

print("¿Tiene duplicados la lista 1?:", verificar_duplicados(lista_con_duplicados))
print("¿Tiene duplicados la lista 2?:", verificar_duplicados(lista_sin_duplicados)) 
````

````python
# Ejercicio 8
# Definición de la función lambda para verificar duplicados en una línea
comprobar_duplicados_lambda = lambda lista: len(lista) != len(set(lista))

# --- PRUEBA DEL EJERCICIO 8 ---
lista_prueba_lambda = [1, 2, 3, 2, 4]
print("¿Tiene duplicados? (Lambda):", comprobar_duplicados_lambda(lista_prueba_lambda))
````

````python
# Ejercicio 9
# Definición de la función para verificar si es palíndromo
def es_palindromo(cadena):
    # Normalizamos el texto: pasamos a minúsculas y removemos espacios en blanco
    cadena_limpia = cadena.lower().replace(" ", "")
    # Comparamos la cadena normalizada con su versión invertida
    if cadena_limpia == cadena_limpia[::-1]:
        # Retornamos True si se lee igual en ambos sentidos
        return True
    else:
        # Retornamos False si no cumple con la condición de palíndromo
        return False

# --- PRUEBA DEL EJERCICIO 9 ---
print("¿'reconocer' es palíndromo?:", es_palindromo("reconocer")) 
print("¿'python' es palíndromo?:", es_palindromo("python"))      
````
