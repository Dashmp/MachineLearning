## 1 Acceso a ficheros
### 1.1 Ficheros
* File es un tipo de objeto predefinido en Python (built-in).
* Permite acceder a ficheros desde programas Python.
* Los ficheros son de un tipo especial:
    * Son built-in, pero no son ni números, ni secuencias, ni mappings. Tampoco responden a
operadores en expresiones.
* La función open permite crear objetos de tipo fichero.
Formato general para abrir un fichero:
```text
afile = open(filename, mode)
```

* *mode* es opcional. Por defecto, los ficheros se abren en modo lectura.
* Los datos leidos de un fichero siempre se obtienen en formato string. Lo mismo ocurre con
escritura.
* Los ficheros se deben cerrar invocando close (liberación de recursos).

````python
# Lectura desde fichero usando método 'read'. Devuelve todo el contenido del␣fichero.
mi_fichero = open('res/multiple_lines.txt')
print(mi_fichero.read())
mi_fichero.close()
````

````python
# Lectura linea a linea a través del bucle 'for'.
mi_fichero = open('res/multiple_lines.txt')
for linea in mi_fichero:
    print(f"{linea}")
mi_fichero.close()
````

````python
# leer de un archivo a una lista.
mi_fichero = open('res/multiple_lines.txt')
lineas = mi_fichero.readlines()
mi_fichero.close()
print(lineas)
````

* *close* autómatico con sentencia *with*. Esta es la forma habitual de leer de fichero en Python, context manager


````python
with open(ruta) as mi_fichero:
    for linea in mi_fichero:
        print(linea, end= '')
````

* Se puede abrir varios ficheros en un mismo with.


````python
#abrir varios ficheros en el mismo with
ruta1 = os.path.join("res", "one_line.txt")
ruta2 = os.path.join("res", "multiple_lines.txt")
with open(ruta1) as fichero1, open(ruta2) as fichero2:
    print(fichero1.readlines())
    print(fichero2.readlines())
````

## 1.2 Modos de acceso
* Al crear un objeto de tipo File se puede espeficiar el modo de acceso (lectura/escritura).

| Modo Acceso | Descripción |
| :--- | :--- |
| `r` | Solo Lectura |
| `w` | Solo Escritura (Borra si el archivo ya existe) |
| `x` | Solo Escritura (Falla si el archivo ya existe) |
| `a` | Crea Fichero (Si existe lo abre y se añade al final) |
| `r+` | Lectura y Escritura |
| `b` | Se puede añadir a otros modos para acceso binario |
| `t` | Acceso para archivos de texto (default) |

## 1.3 Acceso para escritura

````python
def crear_lista(tamanyo):
    lista = []
    for i in range(tamanyo):
        lista.append(str(i) + '\n')
    return lista
    
ruta = os.path.join("res", "a_dummy.txt")
with open(ruta, 'wt') as fichero:
    fichero.write('Cabecera del ciclo del for en el ejemplo\n')
    lista = crear_lista(10)
    fichero.writelines(lista)
    print("archivo creado")
````

## Buffering
* Por defecto, el texto que transfieres desde tu programa a un fichero no se guarda en disco
inmediatamente. Se almacena en un buffer.
* Acciones como cerrar un fichero o invocar el método flush fuerzan que se transfiera el con￾tenido del buffer a disco.


````python
ruta = os.path.join("res", "FicheroParaEscritura.txt")
fichero_write = open(ruta, 'w')
fichero_write.write('foter')

fichero_read = open(ruta, "r")
print(fichero_read.readlines())

fichero_write.flush()

print(fichero_read.readlines())

fichero_write.close()
fichero_read.close()
````

## 1.4 Archivos CSV
* Python permite leer datos de ficheros CSV y también escribir ficheros en este formato.
* Popular formato en ciencia de datos.

````python
# tabla_operaciones.csv contiene valores separados por comas
import csv
ruta = os.path.join("res", "tabla_operaciones.csv")
with open(ruta) as fichero:
    data_reader = csv.reader(fichero, delimiter=',')
    for linea in data_reader:
        print(linea[0] + ' ---- ' + linea[1])
````

````python
ruta = os.path.join("res", "tabla_operaciones.csv")
f = open(ruta)
data_reader = csv.reader(f, delimiter = ',')

ruta_o = os.path.join("res", "4_tabla_operaciones.csv")
with open(ruta_o, 'w') as f2:
    csv_writer = csv.writer(f2, delimiter = '|')
    for line in data_reader:
        print(line)
        csv_writer.writerow(line)
f.close()
````

## 2 Docstrings
* Python permite adjuntar documentación a los objetos e inspeccionarla a través de la línea de
comandos o durante la ejecución del programa.
* Los docstrings se almacenan en el atributo *__doc__* de cada objeto.
* El valor de dicho atributo se puede mostrar por medio de la función help.


````python
help(open)
````

* No es necesario editar este atributo directamente.
* Para asociar un docstring a un objeto basta con escribir el texto (entre triples comillas) al
principio de los modulos, funciones o clases, antes del codigo ejecutable

````python
def funcion_de_prueba():
    """Esta es la documentación de la función de prueba"""
    pass
help(funcion_de_prueba)
````

*  Accediendo al atributo *__doc__* sólo se obtiene el docstring.


````python
print(funcion_de_prueba.__doc__)
````

Como se puede observar, *help* añade información adicional. Por ejemplo, para entidades más
grandes, *help* muestra el docstring dividido en secciones.

````python
import sys
help(sys)
````

Los docstrings que ocupan más de una línea suelen tener estas partes:
* Resumen en la primera línea.
* Línea en blanco.
* Descripción detallada.
* Otra línea en blanco antes del código.


````python
def funcion_de_prueba2():
    """Esta es la línea para el resumen.
    Este es el párrafo donde se puede escribir una descripción más
    detallada de la función. Observa que el resumen y la descripción
    detallada van separados por una línea en blanco. También hay
    otra línea en blanco antes de empezar el código.
    """
    pass
help(funcion_de_prueba2)
````

## 2.4 Ejercicios Resueltos

````python
# =============================================================================
# EJERCICIO 1: Buscar una cadena de caracteres dentro de un archivo de texto
# =============================================================================
import os
RUTA_BASE = "res"
def buscar_cadena_en_fichero(ruta_fichero, cadena_a_buscar):
    # Usamos try-except para gestionar limpiamente posibles errores si el archivo no existe
    try:
        # Abrimos el fichero en modo lectura ('r') con codificación UTF-8 para evitar problemas de acentos[cite: 7]
        with open(ruta_fichero, 'r', encoding='utf-8') as fichero:
            # Recorremos el fichero línea por línea de manera eficiente[cite: 7]
            for linea in fichero:
                # Evaluamos si la palabra o cadena buscada está presente en la línea actual
                if cadena_a_buscar in linea:
                    # Retornamos True inmediatamente al encontrar la primera coincidencia
                    return True
        # Si el ciclo for termina y recorre todo el archivo sin coincidencias, retorna False
        return False
    # Capturamos la excepción específica si el archivo de la carpeta no se encuentra disponible
    except FileNotFoundError:
        # Notificamos al usuario de manera descriptiva el problema de localización
        print(f"Error: No se pudo localizar el archivo en la ruta: '{ruta_fichero}'")
        # Retornamos un valor por defecto seguro para que tu flujo en Jupyter Lab no se detenga
        return False

# --- Bloque de Prueba Real (Ejercicio 1) ---
# Usamos el archivo 'multiple_lines.txt' visible en tu carpeta de capturas de pantalla
ruta_prueba_1 = os.path.join(RUTA_BASE, "multiple_lines.txt")

print("--- Prueba Ejercicio 1 ---")
# Probamos buscando una palabra que podría existir o testeamos la captura del método
print(f"¿Se encuentra la palabra 'linea' en multiple_lines.txt?: {buscar_cadena_en_fichero(ruta_prueba_1, 'linea')}")
````

````python
# =============================================================================
# EJERCICIO 2: Escribir los primeros n elementos de una lista en un fichero destino
# =============================================================================
import os
RUTA_BASE = "res"
def guardar_n_elementos(lista, ruta_destino, n):
    # Controlamos el límite inferior del rango: si n es negativo o cero, escribirá un archivo vacío
    if n <= 0:
        n = 0
    # Controlamos el límite superior del rango: si n excede el tamaño de la lista, lo limitamos al máximo[cite: 7]
    elif n > len(lista):
        # Lanzamos una advertencia clara en la consola para notificar la corrección del rango[cite: 7]
        print(f"Advertencia: 'n' ({n}) excede el tamaño de la lista. Ajustando al máximo de elementos ({len(lista)}).")
        # Reasignamos n para igualarlo a la longitud real de la lista y evitar un IndexError[cite: 7]
        n = len(lista)
        
    try:
        # Abrimos o creamos el archivo en modo escritura ('w') dentro de la ruta especificada[cite: 7]
        with open(ruta_destino, 'w', encoding='utf-8') as fichero:
            # Iteramos con un ciclo indexado desde 0 hasta el valor de n corregido
            for i in range(n):
                # Escribimos el elemento convirtiéndolo a texto plano más un salto de línea[cite: 7]
                fichero.write(str(lista[i]) + '\n')
        # Imprimimos en tu cuaderno de Jupyter la confirmación del guardado correcto
        print(f"Éxito: Archivo guardado correctamente en '{ruta_destino}' con {n} líneas.")
    # Atajamos cualquier inconveniente de permisos del sistema operativo o rutas mal estructuradas
    except Exception as e:
        # Mostramos la descripción exacta del error capturado de forma controlada
        print(f"Ocurrió un error inesperado al intentar guardar el archivo: {e}")

# --- Bloque de Prueba Real (Ejercicio 2) ---
# Definimos una lista de compras de prueba
elementos_prueba = ["Python", "JupyterLab", "Data Science", "Pandas", "Numpy"]
# Usamos 'a_dummy.txt' o 'e_dummy.txt', que son nombres de archivos que ya manejas en tu carpeta res
ruta_prueba_2 = os.path.join(RUTA_BASE, "a_dummy.txt")

print("\n--- Prueba Ejercicio 2 ---")
# Forzamos intencionalmente un valor de n fuera de rango (15) para probar tu manejador de límites
guardar_n_elementos(elementos_prueba, ruta_prueba_2, 15)
````

````python
# =============================================================================
# EJERCICIO 3: Calcular la frecuencia de aparición de cada palabra en un fichero
# =============================================================================
import os
RUTA_BASE = "res"
def calcular_frecuencia_palabras(ruta_fichero):
    # Creamos un diccionario vacío donde guardaremos el mapeo de "palabra": conteo[cite: 7]
    frecuencias = {}
    try:
        # Abrimos el archivo de texto en modo lectura estándar[cite: 7]
        with open(ruta_fichero, 'r', encoding='utf-8') as fichero:
            # Leemos por completo el bloque de texto del archivo y lo guardamos como un solo string
            contenido = fichero.read()
            # Pasamos todo el string a minúsculas para unificar y llamamos a split() para romperlo en palabras[cite: 7]
            palabras = contenido.lower().split()
            
            # Iteramos secuencialmente sobre la lista de palabras generadas[cite: 7]
            for palabra in palabras:
                # Removemos caracteres especiales de puntuación que suelen quedar pegados a las palabras
                palabra = palabra.strip(".,;:¡!¿?()\"'")
                # Validamos que el string resultante no esté vacío tras la limpieza de caracteres
                if palabra:
                    # Si la palabra ya está registrada en nuestro diccionario, aumentamos su contador[cite: 7]
                    if palabra in frecuencias:
                        frecuencias[palabra] += 1
                    # Si es la primera vez que mapeamos la palabra, la inicializamos con valor 1[cite: 7]
                    else:
                        frecuencias[palabra] = 1
        # Retornamos el mapa de frecuencias estructurado al finalizar el procesamiento[cite: 7]
        return frecuencias
    # Si intentas leer un archivo que no está en la carpeta de recursos, capturamos el fallo
    except FileNotFoundError:
        # Imprimimos una alerta limpia en la celda de Jupyter
        print(f"Error: No se pudo procesar las frecuencias. Archivo inexistente: '{ruta_fichero}'")
        # Devolvemos una estructura vacía consistente para mantener la estabilidad del software[cite: 7]
        return {}

# --- Bloque de Prueba Real (Ejercicio 3) ---
# Usamos 'one_line.txt' que se encuentra en la raíz de tu estructura res mostrada
ruta_prueba_3 = os.path.join(RUTA_BASE, "one_line.txt")

print("\n--- Prueba Ejercicio 3 ---")
resultado_frecuencias = calcular_frecuencia_palabras(ruta_prueba_3)
print(f"Frecuencias de palabras encontradas en {ruta_prueba_3}:")
print(resultado_frecuencias)
````
