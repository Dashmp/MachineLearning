# 6 frame.plot()
frame.plot(x,y,label=<Series>,ax,style,kind,xticks,yticks,title,subplots)
* (solo para Series) label: referencia para la leyenda
* ax: subfigura en la que dibujar los datos
* style: estilo de la línea
* kind: tipo de gráfico (’bar’, ‘pie’, ‘hist’, ‘area’ ‘line’, ‘barh’, ‘density’, ‘kde’).
* xticks y yticks: valores en los ejes X e Y
* title: string como título
* (solo para DataFrame) subplots: bool si se desean subfiguras separadas para cada columna.
* (solo para DataFrame) ‘x’ e ‘y’ se pueden utilizar para seleccionar columnas para cada eje

```python
#scatter de clases, coloreando segun clase en matplotlib
iris = sns.load_dataset('iris')

# preparacion de los datos
data = []
colors = ["red", "green", "blue"]
groups = ["setosa", "versicolor", "virginica"]
columns = ["petal_length", "petal_width"]
for i in groups:
    # busqueda por clase definida
    data.append(iris[iris["species"] == i][columns].values)
    
#creacion de la grafica
fig = plt.figure()
ax = fig.add_subplot()
for count, d in enumerate(data):
    # separar las variables del grafico
    x, y = d[:,0], d[:,1]
    ax.scatter(x, y, alpha = 0.4, c = colors[count -1], edgecolors = 'none', s= 30, label = groups[count - 1])

#etiquetas
ax.set_xlabel(columns[0])
ax.set_ylabel(columns[1])
plt.legend(loc='best')
plt.title('Matplot scatter plot')
plt.show()
```

```python
# =============================================================================
# PREPARACIÓN DE DATOS EJERCICIOS MATPLOTLIB
# =============================================================================

# Crear la lista de años de 2000 a 2020 para el eje X
anos = list(range(2000, 2021))

# Fijamos una semilla aleatoria para que los datos no cambien en cada ejecución
np.random.seed(42)

# Generar 4 secuencias de datos aleatorios uniformes (un valor por cada año, longitud 21)
# Simulamos datos de GDP escalados entre 100 y 500 para que tengan coherencia visual
y0 = np.random.uniform(100, 300, len(anos))
y1 = np.random.uniform(150, 400, len(anos))
y2 = np.random.uniform(200, 450, len(anos))
y3 = np.random.uniform(250, 500, len(anos))

# Guardamos los nombres y los datos en listas para poder iterar cómodamente sobre ellos
paises = ["Hawaii", "San Marino", "Islas Feroe", "Guayana"]
datos_y = [y0, y1, y2, y3]
colores = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"]
```

```python
# =============================================================================
# PARTE 1: TODAS LAS SECUENCIAS EN UNA SOLA FIGURA
# =============================================================================

# Inicializamos la primera figura independiente
plt.figure(figsize=(10, 5))

# Iteramos para graficar cada país con su respectiva etiqueta y color personalizado
for i in range(4):
    plt.plot(anos, datos_y[i], marker="o", label=paises[i], color=colores[i], linewidth=2)

# Añadimos la información requerida
plt.title("Datos aleatorios: Comparativa Global de GDP", fontsize=14, fontweight="bold")
plt.xlabel("Año", fontsize=11)
plt.ylabel("GDP", fontsize=11)
plt.xticks(range(2000, 2021, 2))  # Ajustamos los saltos del eje X cada 2 años
plt.grid(True, linestyle="--", alpha=0.5)

# Añadimos la leyenda para identificar las secuencias
plt.legend(loc="upper left")

plt.show()
```

```python
# =============================================================================
# PARTE 2: CUATRO SUBFIGURAS (AXES) DISTINTAS + ANOTACIÓN
# =============================================================================

# Creamos una cuadrícula de 2x2 subplots compartiendo la misma figura
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(12, 8), sharex=True)

# Aplanamos la matriz de "axes" (de 2x2 a un array de 4 elementos) para iterar fácilmente
axes_flat = axes.flatten()

# Dibujamos cada país en su respectivo subgráfico
for i in range(4):
    ax = axes_flat[i]
    ax.plot(anos, datos_y[i], marker="s", color=colores[i], label=paises[i])

    # Información obligatoria por cada subfigura
    ax.set_title(f"Datos aleatorios - {paises[i]}", fontsize=11)
    ax.set_xlabel("Año")
    ax.set_ylabel("GDP")
    ax.grid(True, linestyle=":", alpha=0.6)
    ax.legend(loc="lower left")

# --- Añadir una anotación con texto y flecha al primer gráfico (Hawaii) ---
# Buscamos el índice del valor máximo en y0 para apuntar directamente a él
idx_max = np.argmax(y0)
ano_max = anos[idx_max]
gdp_max = y0[idx_max]

axes_flat[0].annotate(
    "Pico Máximo",  # Texto de la anotación
    xy=(ano_max, gdp_max),  # Coordenada (X, Y) a la que apunta la flecha
    xytext=(ano_max - 4, gdp_max - 50),  # Coordenada (X, Y) donde se coloca el texto
    arrowprops=dict(
        facecolor="black", shrink=0.08, width=1, headwidth=6
    ),  # Estilo de la flecha
    fontweight="bold",
    color="darkred",
)

# Ajustamos la distribución de los subplots automáticamente para evitar solapamientos
plt.tight_layout()
plt.show()
```

```python
# =============================================================================
# 1. RELACIÓN VISUAL DE CADA PAREJA DE ATRIBUTOS
# =============================================================================
# pairplot genera una matriz de gráficos cruzando todas las variables numéricas
# Usamos 'corner=True' para no duplicar gráficos simétricos y optimizar espacio
sns.pairplot(df_seaborn, corner=True)
plt.suptitle("Matriz de Relaciones entre Atributos (Pairplot)", y=1.02, fontsize=14)
plt.show()
```

```python
# =============================================================================
# 2. CORRELACIÓN ENTRE UNA PAREJA DE VARIABLES CONTINUAS
# =============================================================================
plt.figure(figsize=(7, 5))
# Escogemos 'weight' (peso) y 'displacement' (desplazamiento)
# jointplot muestra el scatter y los histogramas marginales al mismo tiempo
sns.jointplot(data=df_seaborn, x="weight", y="displacement", kind="reg", color="purple")
plt.show()
```

```python
# =============================================================================
# 3. DOS VARIABLES NUMÉRICAS EN BASE A OTRA CATEGÓRICA
# =============================================================================
plt.figure(figsize=(9, 5))
# X: 'horsepower', Y: 'mpg', segmentado cromáticamente por la variable categórica 'origin'
sns.scatterplot(data=df_seaborn, x="horsepower", y="mpg", hue="origin", palette="Set1", alpha=0.8)
plt.title("Relación Potencia vs Consumo según el Origen del Vehículo")
plt.xlabel("Potencia (Horsepower)")
plt.ylabel("Millas por Galón (MPG)")
plt.show()
```

```python
# =============================================================================
# 4. DISTRIBUCIÓN DE UNA VARIABLE NUMÉRICA CONTINUA (Media, outliers, etc.)
# =============================================================================
plt.figure(figsize=(8, 4))
# Combinamos un Boxplot (para ver cuartiles y outliers) con un Stripplot (puntos reales)
sns.boxplot(data=df_seaborn, x="acceleration", color="lightblue", fliersize=5)
sns.stripplot(data=df_seaborn, x="acceleration", color="blue", alpha=0.3, jitter=0.2)
plt.title("Distribución de la Aceleración con Identificación de Outliers")
plt.xlabel("Aceleración")
plt.show()
```

```python
# =============================================================================
# 5. COMPARAR CUANTITATIVAMENTE LA DENSIDAD DE DOS VARIABLES
# =============================================================================
plt.figure(figsize=(9, 5))
# Usamos un gráfico de densidad Kernel (KDE) para superponer las curvas de 'mpg' y 'acceleration'
sns.kdeplot(data=df_seaborn["mpg"], fill=True, label="MPG (Consumo)", color="g", alpha=0.4)
sns.kdeplot(
    data=df_seaborn["acceleration"], fill=True, label="Aceleración", color="m", alpha=0.4
)
plt.title("Comparativa de Densidades Cuantitativas")
plt.xlabel("Valor de la Variable")
plt.ylabel("Densidad")
plt.legend()
plt.show()
```

```python
# =============================================================================
# 6. REGRESIÓN LINEAL DE ORDEN > 1 (POLINOMIAL vs LINEAL ESTÁNDAR)
# =============================================================================
# La relación entre 'horsepower' (potencia) y 'mpg' (consumo) es marcadamente curva (no-lineal).
# Al bajar la potencia, el rendimiento sube de forma exponencial, no recta.

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

# Gráfico Izquierdo: Modelo de regresión lineal simple (order=1) -> Mal ajuste
sns.regplot(data=df_seaborn, x="horsepower", y="mpg", order=1, ax=ax1, color="red")
ax1.set_title("Regresión Lineal Simple (Orden = 1) - Ajuste Deficiente")
ax1.set_xlabel("Potencia")
ax1.set_ylabel("MPG")

# Gráfico Derecho: Modelo de regresión polinomial (order=2) -> Se adapta a la curva perfectamente
sns.regplot(
    data=df_seaborn,
    x="horsepower",
    y="mpg",
    order=2,
    ax=ax2,
    color="dodgerblue",
    line_kws={"color": "darkblue"},
)
ax2.set_title("Regresión Polinomial (Orden = 2) - Ajuste Óptimo")
ax2.set_xlabel("Potencia")

plt.tight_layout()
plt.show()
```
